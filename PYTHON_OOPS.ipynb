# Object-Oriented Programming (OOP) in Python

Object-Oriented Programming (OOP) is a programming style that organizes code around **data (objects)** and **blueprints (classes)**, rather than just writing a list of sequential actions.

---

### 1. What is a Class?
A **Class** is a blueprint, template, or prototype for creating objects. It defines the attributes (data) and methods (behaviors) that the objects created from it will have. 

* Think of a Class as a **house blueprint**. The blueprint itself isn't a house, but it describes exactly how to build one.

---

## 2. What is an Object?
An **Object** is an instance of a class. When a class is defined, no memory is allocated until an object of that class is created.

* If the Class is the house blueprint, the **Object is the actual physical house** built using that blueprint. You can build many individual houses (objects) from a single blueprint (class).

---

## 3. Code Example: Classes and Objects
Here is how you write and execute this concept inside a Python cell:

```python
# Defining the Class (The Blueprint)
class Car:
    # The Constructor method to initialize attributes
    def __init__(self, brand, model, year):
        self.brand = brand   # Attribute
        self.model = model   # Attribute
        self.year = year     # Attribute

    # A Method (Action the object can perform)
    def start_engine(self):
        return f"The {self.year} {self.brand} {self.model}'s engine is now roaring!"

# Creating Objects (Instances of the Class)
car1 = Car("Tesla", "Model S", 2026)
car2 = Car("Ford", "Mustang", 1969)

# Accessing object attributes and methods
print(car1.start_engine())
print(f"Car 2 is a classic {car2.model}.")
```


## PYTHON OOPS CONCEPTS WITH CODING EXAMPLES ##

````python
# 🏗️ THE CLASS (Blueprint)
class Smartphone:
    def __init__(self, brand, model, storage):
        # Attributes
        self.brand = brand
        self.model = model
        self.storage = storage

    # Method (Behavior)
    def display_info(self):
        return f"Device: {self.brand} {self.model} with {self.storage}GB"

# 📱 THE OBJECTS (Real instances created from the blueprint)
phone1 = Smartphone("Apple", "iPhone 17", 256)
phone2 = Smartphone("Samsung", "Galaxy S26", 512)

# Accessing properties and behaviors
print(phone1.display_info())
print(phone2.display_info())
````

````python
# 🔒 ENCAPSULATION EXAMPLE
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # Private attribute: cannot be accessed directly outside the class

    # Getter method to safely read private data
    def get_balance(self):
        return self.__balance

    # Setter method to safely update private data with validation rules
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Successfully deposited ${amount}")
        else:
            print("Invalid deposit amount!")

# Testing the encapsulation
account = BankAccount("Alex", 1000)

# print(account.__balance) # ❌ This line will CRASH because __balance is hidden!
account.deposit(500)       #  Safe modification via a method
print(f"Current Balance: ${account.get_balance()}") #  Safe read via a method
````

````python
# 🧬 INHERITANCE EXAMPLE

# Parent Class (General)
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def log_in(self):
        return f"{self.name} (ID: {self.emp_id}) logged into the system."

# Child Class (Specific) - Inherits from Employee
class Developer(Employee):
    def __init__(self, name, emp_id, programming_language):
        super().__init__(name, emp_id) # Pulls name and emp_id logic from Parent
        self.programming_language = programming_language # Unique attribute for child

    def write_code(self):
        return f"{self.name} is writing clean code in {self.programming_language}."

# Testing the inheritance
dev = Developer("Sarah", "DEV-404", "Python")

print(dev.log_in())     #  Inherited from Parent
print(dev.write_code()) #  Unique to Child
````

````python
# 🌫️ ABSTRACTION EXAMPLE
from abc import ABC, abstractmethod

# Abstract Base Class (The strict template)
class PaymentProcessor(ABC):
    
    @abstractmethod
    def process_payment(self, amount):
        """Every payment method MUST create its own version of this function"""
        pass

# Concrete Class 1
class PayPal(PaymentProcessor):
    def process_payment(self, amount):
        return f"Connecting to PayPal API... Processing payment of ${amount}."

# Concrete Class 2
class CreditCard(PaymentProcessor):
    def process_payment(self, amount):
        return f"Contacting bank gateway... Charging credit card for ${amount}."

# Testing the abstraction
# payment = PaymentProcessor() # ❌ This will CRASH! You cannot create a generic payment object.

user_checkout_1 = PayPal()
user_checkout_2 = CreditCard()

print(user_checkout_1.process_payment(45.00))
print(user_checkout_2.process_payment(120.50))
````

# The SOLID Principles

The **SOLID principles** are a set of five fundamental design principles in object-oriented programming. They help developers write **maintainable, scalable, and flexible code** that is easy to test and extend.

---

### 1. Single Responsibility Principle (SRP)
> *"A class should have only one reason to change."*
*   **Meaning:** A class should do exactly **one job**. If a class has multiple responsibilities, changes to one feature might break another.
*   **Analogy:** A professional chef should only focus on cooking. If they also have to manage the restaurant's accounting and fix the plumbing, they will eventually make mistakes.

### 2. Open/Closed Principle (OCP)
> *"Software entities should be open for extension, but closed for modification."*
*   **Meaning:** You should be able to **add new features** without changing the existing, already-tested code.
*   **Analogy:** A smartphone is open for extension via apps. You don't need to open the hardware and rewire the phone just to install a new game.

### 3. Liskov Substitution Principle (LSP)
> *"Subtypes must be substitutable for their base types."*
*   **Meaning:** A subclass should be able to replace its parent class **without breaking the application**. The child class must honor the behavior expected of the parent.
*   **Analogy:** If you replace a mechanical clock's battery with a solar battery, the clock should still tell the time correctly. If the new battery changes how the clock hands move, it breaks the system.

### 4. Interface Segregation Principle (ISP)
> *"Clients should not be forced to depend upon interfaces that they do not use."*
*   **Meaning:** It is better to have **many small, specific interfaces** rather than one giant, "do-it-all" interface. 
*   **Analogy:** A restaurant menu shouldn't force a strict vegetarian to read through a list of steaks just to find a salad. The menu should be broken down into clear, independent sections.

### 5. Dependency Inversion Principle (DIP)
> *"Depend upon abstractions, not concretions."*
*   **Meaning:** High-level logic should not depend directly on low-level details. Both should **depend on interfaces or abstract classes**.
*   **Analogy:** Your wall outlet provides an abstract interface for electricity. You can plug in a lamp, a laptop, or a vacuum cleaner without needing to rewire your house's electrical grid for every device.

---

### 🚀 Why Use SOLID?
*   **Easier Maintenance:** Code changes stay isolated.
*   **Better Testability:** Small, focused classes are easier to test.
*   **Scalability:** Adding new features won't break your old code.

## 1.) SINGLE RESPONSIBILITY PRINCIPLE ##

````python
## Bad example

class User:

    def __init__(self, name, email):
        self.name = name
        self.email = email

    def save_to_database(self):
        print(f"Saving {self.name} to database")

    def send_email(self):
        print(f"Sending email to {self.email}")


user = User("John", "john@example.com")

user.save_to_database()
user.send_email()
````

````python
## Good Example

class User:

    def __init__(self, name, email):
        self.name = name
        self.email = email


class UserRepository:

    def save(self, user):
        print(f"Saving {user.name} to database")


class EmailService:

    def send_email(self, user):
        print(f"Sending email to {user.email}")


user = User("John", "john@example.com")

repository = UserRepository()
repository.save(user)

email_service = EmailService()
email_service.send_email(user)
````

## Liskov Substitution Principle (LSP) ##

````python
# Liskov Substitution Principle (LSP)
#
# BAD EXAMPLE:
# A child class should not change the expected behavior
# of its parent class.

class Bird:

    def fly(self):
        print("Bird is flying")


class Sparrow(Bird):

    def fly(self):
        print("Sparrow is flying")


class Penguin(Bird):

    def fly(self):
        # Penguin cannot fly
        raise Exception("Penguins cannot fly")


def make_bird_fly(bird: Bird):
    bird.fly()


# Sparrow works correctly
sparrow = Sparrow()
make_bird_fly(sparrow)

# Output:
# Sparrow is flying


# Penguin is also a Bird,
# so according to the parent class, it should be usable here.
penguin = Penguin()
make_bird_fly(penguin)

# ERROR:
# Exception: Penguins cannot fly
````

````python
## Good Example

# Liskov Substitution Principle (LSP)
#
# Definition:
# Objects of a child class should be replaceable
# with objects of the parent class without breaking the program.

class Bird:
    def eat(self):
        print("Bird is eating")


class FlyingBird(Bird):
    def fly(self):
        print("Bird is flying")


class Sparrow(FlyingBird):
    def fly(self):
        print("Sparrow is flying")


class Penguin(Bird):
    def swim(self):
        print("Penguin is swimming")


def make_bird_fly(bird: FlyingBird):
    bird.fly()


# Sparrow can be substituted for FlyingBird
sparrow = Sparrow()
make_bird_fly(sparrow)

# Output:
# Sparrow is flying


# Penguin is a Bird, but it is NOT a FlyingBird.
# Therefore, we don't pass Penguin to make_bird_fly().
penguin = Penguin()
penguin.eat()
penguin.swim()

# Output:
# Bird is eating
# Penguin is swimming
````

##  Interface Segregation Principle (ISP) ##

````python
# Interface Segregation Principle (ISP)
#
# BAD EXAMPLE:
# A class should not be forced to implement methods
# that it does not need.

from abc import ABC, abstractmethod


class Machine(ABC):

    @abstractmethod
    def print_document(self):
        pass

    @abstractmethod
    def scan_document(self):
        pass

    @abstractmethod
    def fax_document(self):
        pass


class Printer(Machine):

    def print_document(self):
        print("Printing document")

    def scan_document(self):
        # Printer cannot scan
        raise Exception("Printer cannot scan")

    def fax_document(self):
        # Printer cannot fax
        raise Exception("Printer cannot fax")


printer = Printer()

printer.print_document()

# These methods should not be forced on Printer
printer.scan_document()
printer.fax_document()
````

````python
## Good Example

from abc import ABC, abstractmethod


class Printer(ABC):

    @abstractmethod
    def print_document(self):
        pass


class Scanner(ABC):

    @abstractmethod
    def scan_document(self):
        pass


class Fax(ABC):

    @abstractmethod
    def fax_document(self):
        pass


class SimplePrinter(Printer):

    def print_document(self):
        print("Printing document")


class MultiFunctionPrinter(Printer, Scanner, Fax):

    def print_document(self):
        print("Printing document")

    def scan_document(self):
        print("Scanning document")

    def fax_document(self):
        print("Faxing document")


# Simple printer only implements what it needs
printer = SimplePrinter()
printer.print_document()

# Multi-function printer supports all operations
mfp = MultiFunctionPrinter()
mfp.print_document()
mfp.scan_document()
mfp.fax_document()
````

##  DIP — Dependency Inversion Principle ##

````python
## Bad Example

class MySQLDatabase:

    def save_user(self, username):
        print(f"Saving {username} to MySQL")


class UserService:

    def __init__(self):
        # Direct dependency on MySQL
        self.database = MySQLDatabase()

    def create_user(self, username):
        self.database.save_user(username)


service = UserService()

service.create_user("John")
````

````python
## Good Example

from abc import ABC, abstractmethod


# Abstraction
class Database(ABC):

    @abstractmethod
    def save_user(self, username):
        pass


# Low-level implementation
class MySQLDatabase(Database):

    def save_user(self, username):
        print(f"Saving {username} to MySQL")


# Another low-level implementation
class PostgreSQLDatabase(Database):

    def save_user(self, username):
        print(f"Saving {username} to PostgreSQL")


# High-level class depends on abstraction
class UserService:

    def __init__(self, database: Database):
        self.database = database

    def create_user(self, username):
        self.database.save_user(username)


# Use MySQL
mysql = MySQLDatabase()
service = UserService(mysql)

service.create_user("John")


# Use PostgreSQL
postgresql = PostgreSQLDatabase()
service = UserService(postgresql)

service.create_user("David")
````
